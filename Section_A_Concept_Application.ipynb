### Question 1
**Question: You open the Heart Disease dataset in Jupyter Notebook. What is the difference between a .py file and a .ipynb file?**

**Answer:**
A `.py` file is a standard Python script containing pure Python code meant to be executed sequentially from start to finish via the command line or an interpreter. It contains no native formatting or rich outputs. A `.ipynb` file (Interactive Python Notebook) is a JSON-based document structure used by Jupyter environments. It allows the interleaving of live executable code cells, rich Markdown text, equations, and inline visual outputs (like graphs or dataframes) within the same document, making it ideal for exploratory data analysis.

### Question 2
**Question: The dataset has columns like age, cholesterol, and sex. What is the difference between a numerical feature and a categorical feature?**

**Answer:**
A **numerical feature** represents quantifiable, measurable data where mathematical operations (like calculating an average) make logical sense. For example, 'age' (e.g., 45, 60) and 'cholesterol' (e.g., 180 mg/dL, 240 mg/dL) are numerical because they represent continuous or discrete quantities. 
A **categorical feature** represents distinct groups, classes, or labels where mathematical calculations are illogical. For example, 'sex' (Male/Female) categorizes data into groups. Even if 'sex' is encoded as 0 and 1, taking the average of 0 and 1 holds no mathematical meaning in reality.

### Question 3
**Question: You run type() on the target column and it returns int. Why does the data type of your target variable matter before framing a prediction problem?**

**Answer:**
The data type of the target variable fundamentally dictates the type of machine learning problem you are solving, which in turn determines the algorithms and evaluation metrics you must use. If the target is an integer representing discrete categories (e.g., 0 for no heart disease, 1 for heart disease), it is a **Classification** problem. If the target is a continuous numerical float (e.g., predicting the exact blood pressure value), it is a **Regression** problem. Applying a regression algorithm to a classification target (or vice versa) will result in structurally invalid outputs and mathematical errors.

### Question 4
**Question: You compute the correlation between age and heart disease presence and get 0.23. What does a correlation value tell you about two variables?**

**Answer:**
A correlation value (which strictly ranges from -1.0 to 1.0) measures the strength and direction of the linear relationship between two numerical variables. 
*   The **sign (+/-)** indicates the direction. A positive 0.23 indicates a positive relationship: as age increases, the presence of heart disease tends to increase.
*   The **magnitude (0.23)** indicates the strength. Because 0.23 is relatively close to 0, it suggests a weak linear relationship. This means while there is a slight upward trend linking higher age to heart disease presence, age alone is not a strong or definitive linear predictor of the disease on its own.

### Question 5
**Question: A hospital rule says: if cholesterol > 240, flag as high risk. What is the fundamental difference between a rule-based system and a data-driven approach?**

**Answer:**
A **rule-based system** relies on explicitly programmed, hardcoded logical thresholds defined in advance by human experts (e.g., manually writing `if cholesterol > 240: return High Risk`). It cannot adapt to new patterns or handle complex interactions between multiple variables without humans writing more rules.

A **data-driven approach** (Machine Learning) does not use hardcoded thresholds. Instead, it is fed historical patient data containing multiple variables (cholesterol, age, max heart rate) and their actual outcomes. The algorithm automatically discovers and learns the complex, hidden mathematical patterns and optimal thresholds on its own, allowing it to adapt and often outperform simplistic human-written rules.
